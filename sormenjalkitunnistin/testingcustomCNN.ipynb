```python
import keras
import numpy as np

model = keras.models.load_model("v5.keras")

img = keras.utils.load_img(
    "2.jpeg",
    target_size=(128, 128)
)
img2 = keras.utils.load_img(
    "5-1.jpeg",
    target_size=(128, 128)
)
img3 = keras.utils.load_img(
    "5-2.jpg",
    target_size=(128, 128)
)

img_array = keras.utils.img_to_array(img)
img_array2 = keras.utils.img_to_array(img2)
img_array3 = keras.utils.img_to_array(img3)

img_array = np.expand_dims(img_array, axis=0)
img_array2 = np.expand_dims(img_array2, axis=0)
img_array3 = np.expand_dims(img_array3, axis=0)

prediction = model.predict(img_array)
prediction2 = model.predict(img_array2)
prediction3 = model.predict(img_array3)

predicted_class = np.argmax(prediction[0])
predicted_class2 = np.argmax(prediction2[0])
predicted_class3 = np.argmax(prediction3[0])

print("Predicted class:", predicted_class)
print("Predicted class:", predicted_class2)
print("Predicted class:", predicted_class3)
```
