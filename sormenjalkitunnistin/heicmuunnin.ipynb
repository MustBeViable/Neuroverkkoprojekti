```python
#ai koodia joka muuttaa .heic kuvat jpg, tensorflow ei hyväksy .heic tiedostoja
from pathlib import Path
from PIL import Image
import pillow_heif

pillow_heif.register_heif_opener()

source = Path("kuvat")

converted = 0

for file in source.rglob("*"):
    if file.is_file() and file.suffix.lower() == ".heic":

        output = file.with_suffix(".jpg")

        try:
            with Image.open(file) as img:
                img.convert("RGB").save(
                    output,
                    "JPEG",
                    quality=85,
                    optimize=True
                )

            # Delete original HEIC
            file.unlink()

            converted += 1
            print(f"Converted: {file} -> {output}")

        except Exception as e:
            print(f"Failed: {file} | {e}")

print(f"\nConversion finished! Converted {converted} images.")

# Check remaining HEIC files
remaining = [
    f for f in source.rglob("*")
    if f.is_file() and f.suffix.lower() == ".heic"
]

print(f"HEIC files remaining: {len(remaining)}")

for file in remaining:
    print(file)
```
